```python
import pandas as pd

train = pd.read_csv("/kaggle/input/competitions/ml2021spring-hw1/covid.train.csv")
test = pd.read_csv("/kaggle/input/competitions/ml2021spring-hw1/covid.test.csv")

print(train.head())
```

```python
# 分离特征和目标
X = train.drop(columns=["tested_positive.2"])
y = train["tested_positive.2"]

# 简单模型
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

print("模型训练完成！")
```

```python
pred = model.predict(test)
```

```python
submission = pd.DataFrame({
    "id": test["id"],
    "tested_positive": pred
})

submission.to_csv("submission.csv", index=False)
print("submission.csv 已保存")
submission.head()
```
